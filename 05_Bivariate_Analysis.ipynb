# Bivariate Analysis

## Titanic Dataset

Bivariate Analysis is the process of studying the relationship between two variables. It helps us understand whether changes or differences in one variable are associated with another variable.

In this notebook, we will analyze:

- Numerical vs Numerical
- Numerical vs Categorical
- Categorical vs Categorical

The main techniques used are:

- Scatter Plot
- Box Plot
- Violin Plot
- Grouped Bar Chart
- Cross Tabulation

The purpose of this analysis is to understand relationships between important variables in the Titanic dataset before moving toward Machine Learning modeling.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Titanic-Dataset.csv")

print("Dataset loaded successfully.")
print("Dataset Shape:", df.shape)

df.head()
```

# 1. Numerical vs Numerical: Age vs Fare

## Concept

Numerical vs Numerical analysis studies the relationship between two numerical variables.

## Example

In a business dataset, income can be compared with spending to understand whether people with higher income tend to spend more.

In the Titanic dataset, we can compare passenger `Age` with `Fare` to examine whether ticket prices show any relationship with passenger age.

## Implementation

We will use a scatter plot to visualize the relationship between `Age` and `Fare`.

## Output

Each point in the plot represents one passenger.

## Interpretation

The position of the points helps us observe whether there is a positive, negative, or weak relationship between age and fare.

## Insight

The scatter plot can show whether passengers of particular age groups tended to pay similar or different ticket fares.

## AI/ML Relevance

Understanding relationships between numerical features helps identify possible correlations and useful patterns for feature selection and Machine Learning.

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="Age", y="Fare", alpha=0.7)

plt.title("Age vs Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

# 2. Correlation between Age and Fare

## Concept

Correlation is a numerical measure that helps us understand the strength and direction of a relationship between two numerical variables.

## Example

A positive correlation means that two variables generally increase together, while a negative correlation means that one variable tends to decrease when the other increases.

## Implementation

We will calculate the correlation between `Age` and `Fare`.

## Output

The result will provide a correlation value between -1 and 1.

## Interpretation

A value close to 0 indicates a weak linear relationship.

## Insight

The correlation helps us support the observations made from the scatter plot.

## AI/ML Relevance

Correlation analysis can help identify relationships between numerical features and detect possible redundant variables.

```python
age_fare_correlation = df["Age"].corr(df["Fare"])

print("Correlation between Age and Fare:")
print(round(age_fare_correlation, 3))
```

# 3. Numerical vs Categorical: Age vs Survived

## Concept

Numerical vs Categorical analysis compares a numerical variable across different categories.

## Example

In a company, employee salary can be compared across different departments.

In the Titanic dataset, passenger age can be compared between passengers who survived and those who did not survive.

## Implementation

We will use a box plot to compare the distribution of `Age` across the `Survived` categories.

## Output

The plot will show the median, quartiles, spread, and possible outliers for each survival group.

## Interpretation

Differences between the two box plots may indicate differences in the age distribution of survivors and non-survivors.

## Insight

This analysis helps us understand whether age may have a relationship with survival.

## AI/ML Relevance

Comparing numerical features across target categories helps identify potentially useful predictive features.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="Survived", y="Age")

plt.title("Age Distribution by Survival")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Age")

plt.show()
```

# 4. Age vs Survived using Violin Plot

## Concept

A violin plot combines information about the distribution and density of a numerical variable across categories.

## Example

It can show whether passenger ages are concentrated differently among survivors and non-survivors.

## Implementation

We will create a violin plot for `Age` grouped by `Survived`.

## Output

The width of the violin represents areas where observations are more concentrated.

## Interpretation

Differences in the shape of the violins can reveal differences in the age distribution between survival groups.

## Insight

The violin plot provides more distribution information than a standard box plot.

## AI/ML Relevance

Understanding how a numerical feature is distributed across target classes can help identify useful predictive patterns.

```python
plt.figure(figsize=(8, 5))

sns.violinplot(data=df, x="Survived", y="Age")

plt.title("Age Distribution by Survival")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Age")

plt.show()
```

# 5. Numerical vs Categorical: Fare vs Passenger Class

## Concept

This analysis compares a numerical variable across different categories.

## Example

In a business dataset, salary can be compared across departments.

In the Titanic dataset, ticket fare can be compared across First, Second, and Third passenger classes.

## Implementation

We will use a box plot to compare the `Fare` distribution across `Pclass`.

## Output

The visualization will show how ticket fares differ among passenger classes.

## Interpretation

Differences in the median and spread of fare values indicate differences between the ticket classes.

## Insight

Passenger class and ticket fare are expected to have a meaningful relationship because higher classes generally represent different travel categories.

## AI/ML Relevance

Understanding relationships between numerical and categorical variables helps identify meaningful features and possible interactions for Machine Learning.

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="Pclass", y="Fare")

plt.title("Fare Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")

plt.show()
```

# 6. Categorical vs Categorical: Sex vs Survived

## Concept

Categorical vs Categorical analysis examines the relationship between two categorical variables.

## Example

In a business dataset, we can compare department with employee promotion status.

In the Titanic dataset, we can compare passenger sex with survival status.

## Implementation

We will use a grouped bar chart and cross tabulation.

## Output

The analysis will show how survival outcomes are distributed across the different sex categories.

## Interpretation

Differences in bar heights may indicate that survival outcomes vary between categories.

## Insight

This analysis helps us understand whether sex has an observable relationship with survival.

## AI/ML Relevance

Categorical relationships can identify important predictive features for classification models.

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Sex", hue="Survived")

plt.title("Survival Distribution by Sex")
plt.xlabel("Sex")
plt.ylabel("Count")
plt.legend(title="Survived", labels=["No", "Yes"])

plt.show()
```

# 7. Cross Tabulation: Sex vs Survived

## Concept

Cross tabulation is a method used to summarize the frequency of combinations between two categorical variables.

## Example

It can show how many male and female passengers survived or did not survive.

## Implementation

We will create a cross tabulation between `Sex` and `Survived`.

## Output

The table will display the frequency of each combination.

## Interpretation

The results help us directly compare survival counts between the two categories.

## Insight

Cross tabulation provides a clear numerical summary of the relationship observed in the grouped bar chart.

## AI/ML Relevance

Cross tabulation is useful for understanding categorical relationships and identifying variables that may contribute to classification predictions.

```python
sex_survival_crosstab = pd.crosstab(df["Sex"], df["Survived"])

print("Cross Tabulation: Sex vs Survived")
print(sex_survival_crosstab)
```

# 8. Categorical vs Categorical: Passenger Class vs Survived

## Concept

This analysis examines the relationship between passenger class and survival status.

## Example

In a retail dataset, product category can be compared with customer purchase status.

Similarly, in the Titanic dataset, passenger class can be compared with whether a passenger survived.

## Implementation

We will create a grouped bar chart for `Pclass` and `Survived`.

## Output

The visualization will display survival counts within each passenger class.

## Interpretation

Differences between the bars may indicate that survival outcomes varied across passenger classes.

## Insight

Passenger class may provide useful information about differences in survival patterns.

## AI/ML Relevance

Passenger class may be a useful feature for predicting survival in a classification model.

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Pclass", hue="Survived")

plt.title("Survival Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.legend(title="Survived", labels=["No", "Yes"])

plt.show()
```

# 9. Cross Tabulation: Passenger Class vs Survived

## Concept

Cross tabulation provides a table showing the frequency of combinations between passenger class and survival status.

## Example

The table can show how many First Class passengers survived compared with Second and Third Class passengers.

## Implementation

We will create a cross tabulation between `Pclass` and `Survived`.

## Output

The output will show survival counts for each passenger class.

## Interpretation

The table makes it easier to compare survival patterns across different classes.

## Insight

Differences in survival counts suggest that passenger class may have a relationship with survival.

## AI/ML Relevance

Understanding this relationship helps determine whether `Pclass` should be considered an important predictive feature.

```python
pclass_survival_crosstab = pd.crosstab(df["Pclass"], df["Survived"])

print("Cross Tabulation: Pclass vs Survived")
print(pclass_survival_crosstab)
```

# 10. Categorical vs Categorical: Embarked vs Survived

## Concept

This analysis studies the relationship between the embarkation port and passenger survival.

## Example

In a business dataset, customer region can be compared with purchase outcome.

Similarly, the Titanic dataset allows us to compare embarkation location with survival status.

## Implementation

We will use a grouped bar chart and cross tabulation.

## Output

The visualization and table will show survival counts across embarkation categories.

## Interpretation

Differences between categories may indicate variations in passenger characteristics or survival outcomes.

## Insight

Embarkation location may indirectly reflect differences in passenger groups or travel conditions.

# 11. Cross Tabulation: Embarked vs Survived

## Concept

Cross tabulation summarizes the frequency relationship between embarkation port and survival status.

## Example

It shows how many passengers from each embarkation category survived or did not survive.

## Implementation

We will create a cross tabulation between `Embarked` and `Survived`.

## Output

The table will display survival counts for each embarkation category.

## Interpretation

The comparison helps identify whether survival patterns differ across embarkation groups.

## Insight

The relationship should be interpreted together with other variables because embarkation location alone may not explain survival outcomes.

## AI/ML Relevance

This feature may provide useful information when combined with other passenger characteristics.

```python
embarked_survival_crosstab = pd.crosstab(
    df["Embarked"],
    df["Survived"]
)

print("Cross Tabulation: Embarked vs Survived")
print(embarked_survival_crosstab)
```
