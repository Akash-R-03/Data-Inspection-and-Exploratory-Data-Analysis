# Distribution Analysis

In this notebook, we analyze the distribution of important numerical features in the Titanic dataset.

The analysis focuses on the shape, center, spread, skewness, kurtosis, and potential outliers in the data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Titanic-Dataset.csv")

df.head()
```

## Normal Distribution

### Concept

A normal distribution is a symmetric, bell-shaped distribution in which most values are concentrated around the center.

### Example

Human heights are often approximately normally distributed, with most people having an average height and fewer people having extremely short or tall heights.

### Implementation

We visualize the Age variable to check whether its distribution is approximately normal.

### Interpretation

A histogram and density curve help us understand whether the data is symmetric and centered around the mean.

### Insight

Real-world data does not always follow a perfect normal distribution.

### AI/ML Relevance

Understanding normality is useful because some statistical techniques and Machine Learning methods work better when numerical features have approximately normal distributions.

```python
plt.figure(figsize=(8, 4))

sns.histplot(df["Age"].dropna(), kde=True)

plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

## Uniform Distribution

### Concept

A uniform distribution occurs when values are distributed relatively evenly across a range.

### Example

In an ideal fair dice experiment, each number from 1 to 6 has an equal probability of occurring.

### Implementation

The following example creates a simple uniform distribution for demonstration.

### Interpretation

The values are spread relatively evenly across the selected range.

### Insight

Real-world datasets may not contain perfectly uniform features.

### AI/ML Relevance

Understanding the type of distribution helps us select appropriate preprocessing and analytical methods.

```python
uniform_data = np.random.uniform(0, 100, 1000)

plt.figure(figsize=(8, 4))

sns.histplot(uniform_data, bins=20)

plt.title("Example of Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

## Symmetric Distribution

### Concept

A symmetric distribution has a similar shape on both sides of its center.

### Example

In a perfectly symmetric distribution, the left and right sides of the histogram appear approximately balanced.

### Implementation

We compare the mean and median of the Age variable.

### Interpretation

When the mean and median are close, the distribution may be approximately symmetric.

### Insight

Symmetry provides useful information about the overall shape of a numerical feature.

### AI/ML Relevance

Symmetric variables may require less transformation than highly skewed variables.

```python
print("Mean Age:", df["Age"].mean())
print("Median Age:", df["Age"].median())
```

## Right-Skewed Distribution

### Concept

A right-skewed distribution has a longer tail on the right side. A small number of large values pull the distribution toward higher values.

### Example

Income is often right-skewed because most people have moderate incomes while a small number have very high incomes.

### Implementation

We analyze the Fare variable because it may contain unusually high ticket prices.

### Interpretation

A positive skewness value indicates a right-skewed distribution.

### Insight

The mean may be higher than the median because of large values in the right tail.

### AI/ML Relevance

Highly right-skewed features may require transformation before using some Machine Learning algorithms.

```python
print("Fare Skewness:", df["Fare"].skew())

plt.figure(figsize=(8, 4))

sns.histplot(df["Fare"], kde=True)

plt.title("Distribution of Fare")
plt.xlabel("Fare")

plt.show()
```

## Left-Skewed Distribution

### Concept

A left-skewed distribution has a longer tail on the left side. A small number of low values pull the distribution toward lower values.

### Example

If most students score high marks but a few students receive very low marks, the distribution may be left-skewed.

### Implementation

We use a simple example to demonstrate a negatively skewed distribution.

### Interpretation

A negative skewness value generally indicates a left-skewed distribution.

### Insight

The median may be more representative than the mean when the distribution is highly skewed.

### AI/ML Relevance

Skewness can affect scaling, transformation, and the behavior of some Machine Learning models.

```python
left_skewed = np.random.beta(5, 2, 1000)

print("Skewness:", pd.Series(left_skewed).skew())

sns.histplot(left_skewed, kde=True)

plt.title("Example of Left-Skewed Distribution")

plt.show()
```

## Skewness

### Concept

Skewness measures the asymmetry of a distribution.

- Skewness close to 0 indicates an approximately symmetric distribution.
- Positive skewness indicates right skewness.
- Negative skewness indicates left skewness.

### Example

A dataset containing a few extremely high values may have positive skewness.

### Implementation

We calculate skewness for important numerical variables.

### Output

The result shows the direction and approximate level of asymmetry.

### Insight

Skewness helps us identify variables that may need transformation.

### AI/ML Relevance

Highly skewed variables can affect statistical analysis and some Machine Learning algorithms.

```python
print("Age Skewness:", df["Age"].skew())
print("Fare Skewness:", df["Fare"].skew())
```

## Kurtosis

### Concept

Kurtosis describes the tail behavior and extremeness of a distribution.

### Example

A distribution with heavier tails may contain more extreme observations than a distribution with lighter tails.

### Implementation

We calculate kurtosis for Age and Fare.

### Interpretation

Kurtosis helps us understand the presence of extreme values and the overall shape of the distribution.

### Insight

Kurtosis should be interpreted together with visualizations and outlier analysis.

### AI/ML Relevance

Features with extreme tails may contain observations that strongly influence some Machine Learning models.

```python
print("Age Kurtosis:", df["Age"].kurt())
print("Fare Kurtosis:", df["Fare"].kurt())
```

## Center and Spread

### Concept

The center of a distribution describes the typical value, while spread describes how widely the values are distributed.

### Example

Mean and median represent the center, while standard deviation and range describe the spread.

### Implementation

We calculate these statistics for Age and Fare.

### Interpretation

A larger spread indicates greater variation in the data.

### Insight

The mean and median together provide a better understanding of the center, especially when the distribution is skewed.

### AI/ML Relevance

Understanding the spread of features is important for scaling and feature preprocessing.

```python
print("Age")
print("Mean:", df["Age"].mean())
print("Median:", df["Age"].median())
print("Standard Deviation:", df["Age"].std())
print("Range:", df["Age"].max() - df["Age"].min())

print("\nFare")
print("Mean:", df["Fare"].mean())
print("Median:", df["Fare"].median())
print("Standard Deviation:", df["Fare"].std())
print("Range:", df["Fare"].max() - df["Fare"].min())
```

## Distribution Analysis of Important Numerical Features

### Concept

A histogram with a density curve helps us analyze the overall shape of a numerical feature.

### Implementation

We visualize Age and Fare to compare their distributions.

### Interpretation

The plots help us observe symmetry, skewness, center, spread, and potential extreme values.

### Insight

Age and Fare may have different distribution patterns, so they may require different preprocessing approaches.

### AI/ML Relevance

Feature distribution analysis helps us make informed decisions about transformation, scaling, and outlier treatment.

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
sns.histplot(df["Age"].dropna(), kde=True)
plt.title("Age Distribution")

plt.subplot(1, 2, 2)
sns.histplot(df["Fare"], kde=True)
plt.title("Fare Distribution")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
sns.boxplot(x=df["Age"])
plt.title("Age Outliers")

plt.subplot(1, 2, 2)
sns.boxplot(x=df["Fare"])
plt.title("Fare Outliers")

plt.tight_layout()
plt.show()
```
