# Univariate Analysis

## Titanic Dataset

Univariate Analysis is the process of analyzing one variable at a time. It helps us understand the distribution, central tendency, variation, frequency, and characteristics of individual variables.

In this notebook, we will analyze:

### Numerical Variables

- Mean
- Median
- Mode
- Minimum
- Maximum
- Range
- Variance
- Standard Deviation
- Quartiles
- Percentiles
- Skewness
- Kurtosis

### Numerical Visualizations

- Histogram
- KDE / Density Plot
- Box Plot

### Categorical Variables

- Frequency
- Unique Values
- Value Counts
- Percentage Distribution

### Categorical Visualizations

- Bar Chart
- Count Plot
- Pie Chart

The purpose of this analysis is to understand each important variable individually before moving to multivariate analysis and Machine Learning.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Titanic-Dataset.csv")

print("Dataset loaded successfully.")
print("Dataset Shape:", df.shape)

df.head()
```

# 1. Numerical Variable: Age

## Concept

The `Age` variable represents the age of each passenger. It is a continuous numerical variable because age can take many possible values within a range.

## Example

For example, passengers may have ages such as 5, 18, 32, or 65 years.

## Implementation

We will calculate the main descriptive statistics of the `Age` variable, including mean, median, mode, minimum, maximum, range, variance, standard deviation, quartiles, percentiles, skewness, and kurtosis.

## Output

The output will provide a statistical summary of passenger ages.

## Interpretation

The mean and median help us understand the central age of passengers, while the minimum and maximum show the overall age range.

## Insight

The distribution of age can help us understand whether the Titanic passengers were concentrated in particular age groups.

## AI/ML Relevance

Age can be an important feature in a Machine Learning model because survival patterns may vary across different age groups.

```python
age_stats = {
    "Mean": df["Age"].mean(),
    "Median": df["Age"].median(),
    "Mode": df["Age"].mode()[0],
    "Minimum": df["Age"].min(),
    "Maximum": df["Age"].max(),
    "Range": df["Age"].max() - df["Age"].min(),
    "Variance": df["Age"].var(),
    "Standard Deviation": df["Age"].std(),
    "25th Percentile": df["Age"].quantile(0.25),
    "50th Percentile": df["Age"].quantile(0.50),
    "75th Percentile": df["Age"].quantile(0.75),
    "Skewness": df["Age"].skew(),
    "Kurtosis": df["Age"].kurt()
}

print(pd.Series(age_stats).round(2))
```

# 2. Age Distribution using Histogram

## Concept

A histogram shows how numerical values are distributed across different intervals.

## Example

Passenger ages can be grouped into intervals such as 0–10, 10–20, 20–30, and so on.

## Implementation

We will create a histogram for the `Age` variable.

## Output

The visualization will show the frequency of passengers across different age groups.

## Interpretation

The tallest bars represent the age ranges containing the highest number of passengers.

## Insight

The histogram helps us understand the concentration and overall distribution of passenger ages.

## AI/ML Relevance

Understanding the distribution of a feature helps us identify skewness, unusual values, and possible preprocessing requirements.

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Age"].dropna(), bins=20, edgecolor="black")

plt.title("Distribution of Passenger Age")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

# 3. Age Distribution using KDE Plot

## Concept

A KDE, or Kernel Density Estimate, is a smooth curve that represents the distribution of numerical data.

## Example

Instead of grouping ages into bars, a KDE plot provides a smooth view of where passenger ages are concentrated.

## Implementation

We will create a density plot for the `Age` variable.

## Output

The curve will show the density of passenger ages.

## Interpretation

Peaks in the curve represent age ranges where more passengers are concentrated.

## Insight

The KDE plot provides a smoother understanding of the age distribution than a histogram alone.

## AI/ML Relevance

Understanding the shape of a numerical distribution can help determine whether transformations or other preprocessing methods may be required.

```python
plt.figure(figsize=(8, 5))

sns.kdeplot(df["Age"].dropna(), fill=True)

plt.title("Age Density Distribution")
plt.xlabel("Age")

plt.show()
```

# 4. Age Distribution using Box Plot

## Concept

A box plot summarizes the distribution of a numerical variable using quartiles, the median, and potential outliers.

## Example

The box represents the middle 50% of passenger ages, while points outside the normal range may represent unusual observations.

## Implementation

We will create a box plot for the `Age` variable.

## Output

The plot will display the median, quartiles, spread, and possible outliers.

## Interpretation

The central line represents the median, while the box shows the interquartile range.

## Insight

The box plot helps identify whether unusual age values are present.

## AI/ML Relevance

Outliers can influence some Machine Learning algorithms and may require further investigation or preprocessing.

```python
plt.figure(figsize=(8, 4))

sns.boxplot(x=df["Age"])

plt.title("Box Plot of Passenger Age")
plt.xlabel("Age")

plt.show()
```

## Output

## Interpretation

Comparing the mean and median can help us understand whether the fare distribution is symmetrical or skewed.

## Insight

Large differences in ticket prices may indicate substantial variation in passenger travel conditions.

## AI/ML Relevance

Fare can be an important predictive feature because it may be related to passenger class and survival probability.

```python
fare_stats = {
    "Mean": df["Fare"].mean(),
    "Median": df["Fare"].median(),
    "Mode": df["Fare"].mode()[0],
    "Minimum": df["Fare"].min(),
    "Maximum": df["Fare"].max(),
    "Range": df["Fare"].max() - df["Fare"].min(),
    "Variance": df["Fare"].var(),
    "Standard Deviation": df["Fare"].std(),
    "25th Percentile": df["Fare"].quantile(0.25),
    "50th Percentile": df["Fare"].quantile(0.50),
    "75th Percentile": df["Fare"].quantile(0.75),
    "Skewness": df["Fare"].skew(),
    "Kurtosis": df["Fare"].kurt()
}

print(pd.Series(fare_stats).round(2))
```

# 9. Categorical Variable: Survived

## Concept

The `Survived` variable indicates whether a passenger survived the Titanic disaster.

- 0 = Did Not Survive
- 1 = Survived

It is a categorical target variable with two classes.

## Example

A passenger with `Survived = 1` survived, while a passenger with `Survived = 0` did not survive.

## Implementation

We will analyze frequency, unique values, value counts, and percentage distribution.

## Output

The results will show the number and percentage of passengers in each survival category.

## Interpretation

The distribution shows whether the target classes are balanced or imbalanced.

## Insight

The survival variable provides the outcome that can potentially be predicted using other passenger features.

## AI/ML Relevance

`Survived` can be used as the target variable in a binary classification Machine Learning problem.

```python
survival_counts = df["Survived"].value_counts()

survival_percentage = (
    df["Survived"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("Unique Values:")
print(df["Survived"].unique())

print("\nFrequency:")
print(survival_counts)

print("\nPercentage Distribution:")
print(survival_percentage)
```

# 10. Survival Distribution using Bar Chart

## Concept

A bar chart displays the frequency of each category.

## Example

The chart compares the number of passengers who survived with the number who did not survive.

## Implementation

We will create a bar chart using the survival counts.

## Output

The bars represent the number of passengers in each survival category.

## Interpretation

The taller bar represents the more frequent outcome.

## Insight

This visualization provides a quick understanding of the target variable distribution.

## AI/ML Relevance

Class distribution is important because an imbalanced target variable may require special treatment during Machine Learning model development.

```python
survival_counts.plot(kind="bar")

plt.title("Survival Distribution")
plt.xlabel("Survived")
plt.ylabel("Frequency")

plt.xticks([0, 1], ["Did Not Survive", "Survived"], rotation=0)

plt.show()
```

# 11. Survival Distribution using Pie Chart

## Concept

A pie chart shows the percentage contribution of each category to the total.

## Example

The chart can show the proportion of passengers who survived and did not survive.

## Implementation

We will visualize the percentage distribution of the `Survived` variable.

## Output

Each section represents the proportion of one survival category.

## Interpretation

The size of each section shows the relative proportion of passengers in that category.

## Insight

The chart provides a simple overall view of the survival distribution.

## AI/ML Relevance

Target class proportions should be examined before training a classification model.

```python
plt.figure(figsize=(6, 6))

survival_counts.plot(
    kind="pie",
    autopct="%1.1f%%",
    labels=["Did Not Survive", "Survived"]
)

plt.title("Survival Percentage Distribution")
plt.ylabel("")

plt.show()
```
