## 14. Frequency Analysis using value_counts()

### Concept

The `value_counts()` function counts how frequently each unique value appears in a column.

### Example

For the `Sex` column, it can show how many passengers were male and female.

### Implementation

We use `value_counts()` on selected categorical columns.

### Output

The output displays the frequency of each category.

### Interpretation

The result helps us understand the distribution and balance of categorical variables.

### Insight

Some categories may occur much more frequently than others.

### AI/ML Relevance

Understanding category frequencies helps identify class imbalance and categorical distribution before Machine Learning.

```python
print("Sex:")
print(df["Sex"].value_counts())

print("\nPclass:")
print(df["Pclass"].value_counts())

print("\nEmbarked:")
print(df["Embarked"].value_counts())

print("\nSurvived:")
print(df["Survived"].value_counts())
```

## 18. Identifier Columns

### Concept

An identifier column is used to uniquely identify individual records.

### Example

In the Titanic dataset, `PassengerId` represents a unique identifier for each passenger.

### Implementation

We inspect the number of unique values in `PassengerId`.

### Output

The output compares the number of unique passenger IDs with the total number of rows.

### Interpretation

If the number of unique values is equal to the number of rows, the column can be considered an identifier.

### Insight

Identifier columns are useful for record identification but may not contain meaningful predictive information.

### AI/ML Relevance

Identifier columns are generally excluded from Machine Learning features unless they contain useful information.

```python
print("Total Rows:", len(df))
print("Unique PassengerId Values:", df["PassengerId"].nunique())

if df["PassengerId"].nunique() == len(df):
    print("PassengerId is an identifier column.")
```
