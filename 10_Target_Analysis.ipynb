# Target Variable Analysis

## Dataset

This notebook uses the Titanic dataset:

`Titanic-Dataset.csv`

## 1. Target Variable Identification

### Concept

A target variable is the variable that a Machine Learning model tries to predict. In the Titanic dataset, the `Survived` column is the target variable.

### Example

If `Survived = 1`, the passenger survived. If `Survived = 0`, the passenger did not survive.

### Interpretation

The target variable has two possible classes, so the Titanic dataset represents a binary classification problem.

### Insight

Identifying the target variable is the first step before building a Machine Learning model.

### AI/ML Relevance

The target variable helps us identify the type of Machine Learning problem and select suitable models and evaluation metrics.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Titanic-Dataset.csv")

target = "Survived"

print("Target Variable:", target)
print("Unique Values:", df[target].unique())
```

## 2. Class Distribution

### Concept

Class distribution shows how the observations are distributed across the different classes of the target variable.

### Example

In the Titanic dataset, we compare the number of passengers who survived with the number of passengers who did not survive.

### Interpretation

This analysis helps us understand whether both target classes are equally represented.

### Insight

A difference in the number of observations between classes may indicate class imbalance.

### AI/ML Relevance

Understanding class distribution is important because an imbalanced target variable can affect Machine Learning model performance.

```python
class_counts = df[target].value_counts()

print("Class Distribution:")
print(class_counts)
```

## 3. Class Counts

### Concept

Class counts show the exact number of observations available in each target class.

### Example

We can count the number of passengers who survived and the number who did not survive.

### Interpretation

The output provides a direct numerical summary of the target variable.

### Insight

The Titanic dataset contains more passengers who did not survive than passengers who survived.

### AI/ML Relevance

Class counts help us determine whether the dataset may require techniques such as class weighting or resampling during model development.

```python
print("Number of Passengers in Each Class:")
print(df[target].value_counts())
```

## 4. Class Percentages

### Concept

Class percentages show the proportion of each target class relative to the total number of observations.

### Example

Instead of only counting survivors, we can calculate the percentage of passengers who survived.

### Interpretation

Percentages provide a clearer understanding of the target distribution.

### Insight

The non-survival class represents a larger percentage of the dataset.

### AI/ML Relevance

Class percentages help us understand the level of imbalance and select suitable evaluation metrics for the model.

```python
class_percentages = df[target].value_counts(normalize=True) * 100

print("Class Percentages:")
print(class_percentages.round(2))
```

## 5. Target Variable Visualization

### Concept

Visualizing the target variable makes the class distribution easier to understand.

### Example

A count plot can compare the number of passengers who survived and did not survive.

### Interpretation

The height of each bar represents the number of observations in that class.

### Insight

The visualization provides a clear picture of the difference between the two target classes.

### AI/ML Relevance

Visualizing the target variable helps identify potential class imbalance before training a Machine Learning model.

```python
sns.countplot(data=df, x=target)

plt.title("Distribution of Survived")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.show()
```

## 6. Class Imbalance

### Concept

Class imbalance occurs when one target class has significantly more observations than another class.

### Example

If a dataset contains 95% of one class and only 5% of another class, it has severe class imbalance.

### Interpretation

In the Titanic dataset, the classes are not perfectly balanced, but both classes have a meaningful number of observations.

### Insight

Accuracy alone may not always provide a complete picture of model performance when the target classes are imbalanced.

### AI/ML Relevance

For imbalanced classification problems, metrics such as precision, recall, F1-score, and ROC-AUC can provide a better evaluation of model performance.

```python
imbalance_ratio = class_counts.max() / class_counts.min()

print("Class Imbalance Ratio:", round(imbalance_ratio, 2))
```

## 7. Target Variable Summary

### Concept

A target variable summary combines the important findings from the target analysis.

### Interpretation

The `Survived` column is the target variable in the Titanic dataset.

- `0` represents passengers who did not survive.
- `1` represents passengers who survived.

The dataset represents a binary classification problem.

### Insight

The target classes are not equally distributed, with non-survivors forming the larger class.

### AI/ML Relevance

Understanding the target variable helps in selecting suitable Machine Learning models, preprocessing techniques, and evaluation metrics.

```python
print("Target Variable Summary")
print("-----------------------")
print("Target Variable:", target)
print("Problem Type: Binary Classification")

print("\nClass Counts:")
print(class_counts)

print("\nClass Percentages:")
print(class_percentages.round(2))
```

# Conclusion

The target variable analysis identified `Survived` as the target variable in the Titanic dataset. Since it contains two classes, the dataset represents a binary classification problem.

The analysis of class counts, percentages, distribution, and class imbalance provides a clear understanding of the target before building a Machine Learning model.

Understanding the target variable is important because it helps us select suitable models, preprocessing techniques, and evaluation metrics for effective Machine Learning analysis.
